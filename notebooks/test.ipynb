```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Wed Nov  8 15:04:09 2023

@author: Sam Becker, Vincent van Rie, Divyansh Srivastava 
"""

from initalise_apep import *
import matplotlib.pyplot as plt
from amuse.units import units, constants

ternary_system = Initialise_apep()

plt.figure()
# plt.plot([0,1,2,3], [0,1,4,9])
for particle in ternary_system:
    plt.scatter(
        particle.position.x.value_in(units.au),
        particle.position.y.value_in(units.au),
        label=particle.name,
    )
plt.legend()
plt.show()

def add_stellar_winds(system):
    for star in system:
        # Create a new stellar wind for the star
        wind = new_stellar_wind(star)
        # Add the wind to the star
        star.wind = wind


# Add stellar winds to the system
add_stellar_winds(ternary_system)


print(ternary_system)
# Plot the positions of the stars and the stellar winds
plt.figure()
for particle in ternary_system:
    plt.scatter(
        particle.position.x.value_in(units.au),
        particle.position.y.value_in(units.au),
        label=particle.name,
    )
    if hasattr(particle, 'wind'):
        plt.scatter(
            particle.wind.position.x.value_in(units.au),
            particle.wind.position.y.value_in(units.au),
            label=f"{particle.name} wind",
        )
plt.legend()
plt.show()
```
